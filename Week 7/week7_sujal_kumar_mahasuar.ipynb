```python
import os
import numpy as np
import faiss

from google.colab import files

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from sentence_transformers import SentenceTransformer

from transformers import pipeline
```

```python
uploaded = files.upload()
```

```python
pdf_file = list(uploaded.keys())[0]

print("Uploaded PDF:", pdf_file)
```

```python
loader = PyPDFLoader(pdf_file)

documents = loader.load()
```

```python
print("Total Pages:", len(documents))
```

```python
text = ""

for page in documents:
    text += page.page_content + "\n"
```

```python
print(text[:2000])
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = text_splitter.split_text(text)
```

```python
print("Total Chunks:", len(chunks))
```

```python
chunks[:5]
```

```python
for i, chunk in enumerate(chunks[:5]):
    print(f"\nChunk {i+1}\n")
    print(chunk)
    print("-" * 80)
```

```python
embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)
```

```python
embeddings = embedding_model.encode(
    chunks,
    convert_to_numpy=True
)

print("Embeddings Shape:", embeddings.shape)
```

```python
dimension = embeddings.shape[1]

print("Embedding Dimension:", dimension)
```

```python
index = faiss.IndexFlatL2(dimension)
```

```python
index.add(embeddings)

print("Total Vectors Stored:", index.ntotal)
```

```python
query = "What is the main topic of this document?"
```

```python
query_embedding = embedding_model.encode(
    [query],
    convert_to_numpy=True
)
```

```python
k = 3

distances, indices = index.search(
    query_embedding,
    k
)
```

```python
print("Distances:")

print(distances)
```

```python
print("Retrieved Chunk Indices:")

print(indices)
```

```python
retrieved_chunks = []

for idx in indices[0]:

    retrieved_chunks.append(
        chunks[idx]
    )
```

```python
for i, chunk in enumerate(retrieved_chunks):

    print(f"\nRetrieved Chunk {i+1}\n")

    print(chunk)

    print("-"*80)
```

```python
generator = pipeline(
    "text-generation",
    model="google/flan-t5-base"
)
```

```python
user_question = input("Enter your question: ")
```

```python
question_embedding = embedding_model.encode(
    [user_question],
    convert_to_numpy=True
)
```

```python
k = 3

distances, indices = index.search(
    question_embedding,
    k
)
```

```python
retrieved_context = ""

for idx in indices[0]:

    retrieved_context += chunks[idx]

    retrieved_context += "\n\n"
```

```python
print("Retrieved Context:\n")

print(retrieved_context)
```

```python
prompt = f"""

Answer the question using only the information provided below.

Context:

{retrieved_context}

Question:

{user_question}

Answer:

"""
```

```python
response = generator(
    prompt,
    max_length=200,
    do_sample=False
)
```

```python
answer = response[0]["generated_text"]

print(answer)
```

```python
print("Question:")

print(user_question)
```

```python
print("Generated Answer:")

print(answer)
```

```python
another_question = input("Ask another question: ")
```

```python
another_embedding = embedding_model.encode(
    [another_question],
    convert_to_numpy=True
)

distances, indices = index.search(
    another_embedding,
    3
)
```

```python
context = ""

for idx in indices[0]:

    context += chunks[idx]

    context += "\n\n"

prompt = f"""

Context:

{context}

Question:

{another_question}

Answer:

"""

response = generator(
    prompt,
    max_length=200,
    do_sample=False
)

print(response[0]["generated_text"])
```

```python
example_questions = [

    "What is the main objective of the document?",

    "Give a short summary of the document.",

    "What are the important topics discussed?"

]
```

```python
for question in example_questions:

    print("="*80)

    print("Question:")

    print(question)

    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    )

    distances, indices = index.search(
        question_embedding,
        3
    )

    context = ""

    for idx in indices[0]:

        context += chunks[idx]

        context += "\n\n"

    prompt = f"""

Context:

{context}

Question:

{question}

Answer:

"""

    response = generator(
        prompt,
        max_length=200,
        do_sample=False
    )

    print("\nAnswer:\n")

    print(response[0]["generated_text"])

    print("\n")
```

```python
print("Retrieved Context Example:\n")

print(context)
```

```python
print("Total Document Chunks :", len(chunks))

print("Embedding Dimension :", embeddings.shape[1])

print("Total Stored Vectors :", index.ntotal)
```
