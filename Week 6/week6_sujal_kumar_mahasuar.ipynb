```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import mnist
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
import numpy as np
import matplotlib.pyplot as plt
print("TensorFlow Version:", tf.__version__)
```

```python
(x_train, _), (x_test, _) = mnist.load_data()
```

```python
print("Training Images:", x_train.shape)
print("Testing Images:", x_test.shape)
```

```python
print("Training Data Type:", x_train.dtype)
print("Testing Data Type:", x_test.dtype)
```

```python
plt.figure(figsize=(10, 4))
for i in range(10):
    plt.subplot(2, 5, i + 1)
    plt.imshow(x_train[i], cmap="gray")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0
```

```python
print("Minimum Pixel Value:", x_train.min())
print("Maximum Pixel Value:", x_train.max())
```

```python
x_train_clean = np.expand_dims(x_train, axis=-1)
x_test_clean = np.expand_dims(x_test, axis=-1)
print(x_train_clean.shape)
print(x_test_clean.shape)
```

```python
noise_factor = 0.4
np.random.seed(42)
x_train_noisy = x_train_clean + noise_factor * np.random.normal(loc=0.0,scale=1.0,size=x_train_clean.shape)
x_test_noisy = x_test_clean + noise_factor * np.random.normal(loc=0.0,scale=1.0,size=x_test_clean.shape)
```

```python
x_train_noisy = np.clip(x_train_noisy, 0.0, 1.0)
x_test_noisy = np.clip(x_test_noisy, 0.0, 1.0)
```

```python
x_train_noisy = x_train_noisy.astype("float32")
x_test_noisy = x_test_noisy.astype("float32")
```

```python
print("Clean Training Shape:", x_train_clean.shape)
print("Noisy Training Shape:", x_train_noisy.shape)
print("Clean Testing Shape:", x_test_clean.shape)
print("Noisy Testing Shape:", x_test_noisy.shape)
```

```python
print("Noisy Minimum:", x_train_noisy.min())
print("Noisy Maximum:", x_train_noisy.max())
```

```python
plt.figure(figsize=(12, 4))
for i in range(10):
    plt.subplot(2, 10, i + 1)
    plt.imshow(x_test_clean[i].squeeze(), cmap="gray")
    plt.axis("off")
    plt.subplot(2, 10, i + 11)
    plt.imshow(x_test_noisy[i].squeeze(), cmap="gray")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
sample_index = 0
plt.figure(figsize=(3, 3))
plt.imshow(x_test_clean[sample_index].squeeze(), cmap="gray")
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(3, 3))
plt.imshow(x_test_noisy[sample_index].squeeze(), cmap="gray")
plt.title("Noisy Image")
plt.axis("off")
plt.show()
```

```python
print("Number of Training Images:", len(x_train_clean))
print("Number of Testing Images:", len(x_test_clean))
```

```python
print("Image Height:", x_train_clean.shape[1])
print("Image Width:", x_train_clean.shape[2])
print("Channels:", x_train_clean.shape[3])
```

```python
input_shape = (28, 28, 1)
print("Autoencoder Input Shape:", input_shape)
print("Noise Factor:", noise_factor)
```

```python
print("MNIST preprocessing and noise generation completed successfully.")
```

```python
encoder_input = layers.Input(shape=input_shape)
```

```python
x = layers.Conv2D(32,(3, 3),activation="relu",padding="same")(encoder_input)
x = layers.MaxPooling2D((2, 2),padding="same")(x)
```

```python
x = layers.Conv2D(64,(3, 3),activation="relu",padding="same")(x)
x = layers.MaxPooling2D((2, 2),padding="same")(x)
```

```python
x = layers.Conv2D(128,(3, 3),activation="relu",padding="same")(x)
```

```python
encoded = layers.MaxPooling2D((2, 2),padding="same",name="bottleneck")(x)
```

```python
encoder = models.Model(encoder_input,encoded,name="Encoder")
encoder.summary()
```

```python
print("Encoder Input Shape:", encoder.input_shape)
print("Bottleneck Shape:", encoder.output_shape)
```

```python
x = layers.Conv2D(128,(3, 3),activation="relu",padding="same")(encoded)
x = layers.UpSampling2D((2, 2))(x)
```

```python
x = layers.Conv2D(64,(3, 3),activation="relu",padding="same")(x)
x = layers.UpSampling2D((2, 2))(x)
```

```python
x = layers.Conv2D(32,(3, 3),activation="relu",padding="same")(x)
x = layers.UpSampling2D((2, 2))(x)
```

```python
x = layers.Conv2D(16,(3, 3),activation="relu",padding="same")(x)
```

```python
decoded = layers.Conv2D(1,(3, 3),activation="sigmoid",padding="same")(x)
```

```python
decoded = layers.Cropping2D(cropping=((2, 2), (2, 2)))(decoded)
```

```python
autoencoder = models.Model(encoder_input,decoded,name="MNIST_Denoising_Autoencoder")
autoencoder.summary()
```

```python
print("Autoencoder Input Shape:", autoencoder.input_shape)
print("Autoencoder Output Shape:", autoencoder.output_shape)
```

```python
autoencoder.compile(optimizer="adam",loss="binary_crossentropy",metrics=["mse"])
```

```python
early_stopping = EarlyStopping(monitor="val_loss",patience=3,restore_best_weights=True)
```

```python
checkpoint = ModelCheckpoint("best_mnist_autoencoder.keras",monitor="val_loss",save_best_only=True)
```

```python
print("Input:", x_train_noisy.shape)
print("Target:", x_train_clean.shape)
print("Model Output:", autoencoder.output_shape)
```

```python
print("Convolutional Autoencoder created and compiled successfully.")
```

```python
history = autoencoder.fit(x_train_noisy,x_train_clean,epochs=20,batch_size=128,shuffle=True,validation_data=(x_test_noisy, x_test_clean),callbacks=[early_stopping, checkpoint],verbose=1)
```

```python
print("Training Completed Successfully")
print("Epochs Trained:", len(history.history["loss"]))
```

```python
test_loss, test_mse = autoencoder.evaluate(x_test_noisy,x_test_clean,verbose=0)
print("Test Loss:", test_loss)
print("Test MSE:", test_mse)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(history.history["loss"], label="Training Loss")
plt.plot(history.history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Autoencoder Training and Validation Loss")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(history.history["mse"], label="Training MSE")
plt.plot(history.history["val_mse"], label="Validation MSE")
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.title("Autoencoder Training and Validation MSE")
plt.legend()
plt.show()
```

```python
denoised_images = autoencoder.predict(x_test_noisy,verbose=1)
print(denoised_images.shape)
```

```python
denoised_images = np.clip(denoised_images,0.0,1.0)
```

```python
sample_index = 0
plt.figure(figsize=(3, 3))
plt.imshow(x_test_clean[sample_index].squeeze(),cmap="gray")
plt.title("Original")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(3, 3))
plt.imshow(x_test_noisy[sample_index].squeeze(),cmap="gray")
plt.title("Noisy")
plt.axis("off")
plt.show()
```

```python
plt.figure(figsize=(3, 3))
plt.imshow(denoised_images[sample_index].squeeze(),cmap="gray")
plt.title("Denoised")
plt.axis("off")
plt.show()
```

```python
n = 10
plt.figure(figsize=(20, 6))
for i in range(n):
    plt.subplot(3, n, i + 1)
    plt.imshow(x_test_clean[i].squeeze(), cmap="gray")
    plt.axis("off")
    plt.subplot(3, n, i + 1 + n)
    plt.imshow(x_test_noisy[i].squeeze(), cmap="gray")
    plt.axis("off")
    plt.subplot(3, n, i + 1 + 2*n)
    plt.imshow(denoised_images[i].squeeze(), cmap="gray")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
original_mse = np.mean((x_test_clean - x_test_noisy) ** 2)
denoised_mse = np.mean((x_test_clean - denoised_images) ** 2)
print("Noisy Image MSE:", original_mse)
print("Denoised Image MSE:", denoised_mse)
```

```python
mse_improvement = ((original_mse - denoised_mse)/ original_mse) * 100
print("MSE Improvement:", mse_improvement, "%")
```

```python
noisy_psnr = tf.image.psnr(x_test_clean,x_test_noisy,max_val=1.0)
denoised_psnr = tf.image.psnr(x_test_clean,denoised_images,max_val=1.0)
print("Average Noisy PSNR:",float(tf.reduce_mean(noisy_psnr)))
print("Average Denoised PSNR:",float(tf.reduce_mean(denoised_psnr)))
```

```python
final_denoised = best_autoencoder.predict(x_test_noisy,verbose=0)
final_denoised = np.clip(final_denoised,0.0,1.0)
n = 10
plt.figure(figsize=(20, 6))
for i in range(n):
    plt.subplot(3, n, i + 1)
    plt.imshow(x_test_clean[i].squeeze(), cmap="gray")
    plt.title("Original")
    plt.axis("off")
    plt.subplot(3, n, i + 1 + n)
    plt.imshow(x_test_noisy[i].squeeze(), cmap="gray")
    plt.title("Noisy")
    plt.axis("off")
    plt.subplot(3, n, i + 1 + 2*n)
    plt.imshow(final_denoised[i].squeeze(), cmap="gray")
    plt.title("Denoised")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
noisy_mse = np.mean((x_test_clean - x_test_noisy) ** 2)
final_mse = np.mean((x_test_clean - final_denoised) ** 2)
print("Noisy MSE:", noisy_mse)
print("Denoised MSE:", final_mse)
```

```python
improvement = ((noisy_mse - final_mse)/ noisy_mse) * 100
print("MSE Improvement:", round(improvement, 2), "%")
```

```python
noisy_psnr_final = tf.image.psnr(x_test_clean,x_test_noisy,max_val=1.0)
denoised_psnr_final = tf.image.psnr(x_test_clean,final_denoised,max_val=1.0)
avg_noisy_psnr = float(tf.reduce_mean(noisy_psnr_final))
avg_denoised_psnr = float(tf.reduce_mean(denoised_psnr_final))
print("Noisy PSNR:", avg_noisy_psnr)
print("Denoised PSNR:", avg_denoised_psnr)
```

```python
metrics = ["MSE", "PSNR"]
noisy_scores = [noisy_mse,avg_noisy_psnr]
denoised_scores = [final_mse,avg_denoised_psnr]
print("Metrics:", metrics)
print("Noisy:", noisy_scores)
print("Denoised:", denoised_scores)
```
